```python
# feature engineering
class PdaysFeatures(BaseEstimator, TransformerMixin):
    def fit(self, X, y=None):
        return self

    def transform(self, X):
        X = X.copy()

        mask = (
            X["pdays"].eq(999)
            & X["previous"].gt(0)
            & X["poutcome"].ne("nonexistent")
        )

        X["pdays_never"] = X["pdays"].eq(999).astype(int)
        X["is_inconsistent"] = mask.astype(int)

        X.loc[X["pdays"].eq(999), "pdays"] = np.nan

        return X


class IQRClipper(BaseEstimator, TransformerMixin):
    def __init__(self, columns, multiplier=1.5):
        self.columns = columns
        self.multiplier = multiplier

    def fit(self, X, y=None):
        X = X.copy()

        self.bounds_ = {}

        for col in self.columns:
            q1 = X[col].quantile(0.25)
            q3 = X[col].quantile(0.75)
            iqr = q3 - q1

            lo = q1 - self.multiplier * iqr
            hi = q3 + self.multiplier * iqr

            self.bounds_[col] = (lo, hi)

        return self

    def transform(self, X):
        X = X.copy()

        for col in self.columns:
            lo, hi = self.bounds_[col]
            X[col] = X[col].clip(lo, hi)

        return X
```

```python
numeric_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler()),
    ('poly', PolynomialFeatures(degree=2, include_bias=False))
])

categorical_pipeline = Pipeline([
    ('ordinal', OrdinalEncoder(handle_unknown='use_encoded_value', unknown_value=-1, encoded_missing_value=-2, dtype=int))
])

preprocessor_clean = ColumnTransformer([
    ('num', numeric_pipeline, numeric_features_clean),
    ('cat', categorical_pipeline, categorical_features)
])

preprocessor = ColumnTransformer([
    ('num', numeric_pipeline, numeric_features),
    ('cat', categorical_pipeline, categorical_features)
])

preprocessor_boosting = ColumnTransformer(
    transformers=[('cat', categorical_pipeline, categorical_features)],
    remainder='passthrough',
)

# фінальна обробка для даних без витоку
feature_engineering_clean = Pipeline([
    ('pdays_features', PdaysFeatures()),
    ('iqr_clipper', IQRClipper(columns=numeric_features_clean)),
    ('preprocessor', preprocessor_clean)
])

# фінальна обробка для даних з витоком
feature_engineering = Pipeline([
    ('pdays_features', PdaysFeatures()),
    ('iqr_clipper', IQRClipper(columns=numeric_features)),
    ('preprocessor', preprocessor)
])

feature_engineering_boosting = Pipeline(
    steps=[
        ('pdays_features', PdaysFeatures()),
        ('iqr_clipper', IQRClipper(columns=numeric_features_clean)),
        ('preprocessor', preprocessor_boosting),
    ]
)
```

```python
def compute_auroc_and_build_roc(pipeline, inputs, targets, name=''):
  # Predict probabilities
  y_pred_proba = pipeline.predict_proba(inputs)[:, 1]

  # Compute ROC curve
  fpr, tpr, thresholds = roc_curve(targets, y_pred_proba, pos_label=1)

  # Compute AUROC
  roc_auc = auc(fpr, tpr)
  print(f'AUROC for {name}: {roc_auc:.2f}')

  # Plot the ROC curve
  plt.figure()
  plt.plot(fpr, tpr, color='darkorange', lw=2, label=f'ROC curve (area = {roc_auc:.2f})')
  plt.plot([0, 1], [0, 1], color='navy', lw=2, linestyle='--')
  plt.xlim([0.0, 1.0])
  plt.ylim([0.0, 1.05])
  plt.xlabel('False Positive Rate')
  plt.ylabel('True Positive Rate')
  plt.title(f'Receiver Operating Characteristic (ROC) Curve for {name}')
  plt.legend(loc="lower right")
  plt.show()
```

```python
def predict_and_plot(targets_predicted, targets, name=''):
    cf = confusion_matrix(targets, targets_predicted, normalize='true')
    plt.figure()
    sns.heatmap(cf, annot=True)
    plt.xlabel('Prediction')
    plt.ylabel('Target')
    plt.title('{} Confusion Matrix'.format(name));

    return targets_predicted
```

```python
results_randomized = []
best_pipelines = {}

for name, exp in models_boost.items():
    boost_pipeline = Pipeline(
        steps=[
            ('fe_and_prep', feature_engineering_boosting),
            ('classifier', exp['model'])
            ]
    )

    search = RandomizedSearchCV(
        estimator=boost_pipeline,
        param_distributions=exp['param_distributions'],
        n_iter=30,
        cv=5,
        scoring='roc_auc',
        random_state=42,
        n_jobs=-1
    )

    search.fit(X_train_clean, y_train)
    best_pipelines[name] = search.best_estimator_
    val_score = search.score(X_val_clean, y_val)

    results_randomized.append(
        {
            'Model': name,
            'Best Train Score (CV ROC-AUC)': search.best_score_,
            'Validation Score (ROC-AUC)': val_score,
            'Best Params': search.best_params_,
        }
    )
```
